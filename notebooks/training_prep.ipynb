```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("/Users/nidhishgupta/Desktop/GridlockChallenge/data/half_processed/cleaned_data.csv")
```

```python
data.isnull().sum()
```

```python
data['timestamp'].unique()
```

```python
time_parts = data['timestamp'].str.split(':', expand=True)
```

```python
data['hour'] = time_parts[0].astype(int)
data['minute'] = time_parts[1].astype(int)
```

```python
data
```

```python
data['minute_of_day'] = data['hour'] * 60 + data['minute']
```

```python
data
```

```python
MINUTES_IN_DAY = 1440.0
data['time_sin'] = np.sin(2 * np.pi * data['minute_of_day'] / MINUTES_IN_DAY)
data['time_cos'] = np.cos(2 * np.pi * data['minute_of_day'] / MINUTES_IN_DAY)
```

```python
data
```

```python
import pygeohash as pgh

# Decode geohash into center latitude and longitude
data['latitude'] = data['geohash'].apply(lambda x: pgh.decode(x)[0])
data['longitude'] = data['geohash'].apply(lambda x: pgh.decode(x)[1])
```

```python
data
```

```python
data = data.sort_values(by=['geohash', 'timestamp']).reset_index(drop=True)

# Create a lag feature for the previous timestamp's demand
data['demand_lag_1'] = data.groupby('geohash')['demand'].shift(1)
data['demand_lag_2'] = data.groupby('geohash')['demand'].shift(2)

# Fill the first few empty rows created by shifting with the median demand
data['demand_lag_1'] = data['demand_lag_1'].fillna(data['demand'].median())
data['demand_lag_2'] = data['demand_lag_2'].fillna(data['demand'].median())
```

```python
data
```

```python
data.isnull().sum()
```

```python
from sklearn.cluster import KMeans

# Group coordinates into clusters (e.g., 20 distinct traffic zones)
coords = data[['latitude', 'longitude']]
kmeans = KMeans(n_clusters=20, random_state=42, n_init=10)

data['traffic_cluster_id'] = kmeans.fit_predict(coords)
```

```python
data
```

```python
data.columns
```

```python
data['traffic_cluster_id'] = data['traffic_cluster_id'].astype('category')
```

```python
data = pd.get_dummies(data, columns=['Weather', 'RoadType','LargeVehicles','Landmarks'], drop_first=True, dtype=int)
```
