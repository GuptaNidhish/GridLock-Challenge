```python
# Best: Fill missing RoadType using the mode of that specific geographic area (geohash)
data['RoadType'] = data.groupby('geohash')['RoadType'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown'))
```

```python
data.isnull().sum()
```

```python
data['temp_bins'] = pd.cut(data['Temperature'], bins=range(int(data['Temperature'].min())-5, int(data['Temperature'].max())+5, 5))
```

```python
data
```

```python
data['Weather'] = data.groupby('temp_bins', dropna=False)['Weather'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Clear')
)

# Drop the temporary bin column when done
data = data.drop(columns=['temp_bins'])
```
